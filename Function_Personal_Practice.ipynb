```python
def add(a, b):
    return a + b

print(add(10, 20))  #Add number
```

```python
def square(num):
    return num * num

print(square(5)) # square find out
```

```python
def even_odd(num):
    if num % 2 == 0:
        return "Even"
    return "Odd"

print(even_odd(7))   # Checking Even or odd
```

```python
def maximum(a, b):
    if a > b:
        return a
    return b

print(maximum(25, 15)) # Finding maximum numbers
```

```python
def greet(name):
    return "Hello, " + name + "!"

print(greet("Tanmay")) 
```

```python
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

print(factorial(5)) # Facatorial number
```

```python
def count_vowels(text):
    count = 0
    for ch in text.lower():
        if ch in "aeiou":
            count += 1
    return count

print(count_vowels("Python")) #count vowel 
```

```python
def reverse(text):
    return text[::-1]

print(reverse("Python")) # Revese string
```

```python
def prime(num):
    if num <= 1:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True

print(prime(13)) # check if its prime number or not 
```

```python
def largest(lst):
    return max(lst)

numbers = [10, 45, 20, 90, 60]
print(largest(numbers)) # Find the largest number
```
