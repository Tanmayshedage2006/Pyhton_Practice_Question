```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd

df = pd.read_csv("industry_retail_orders_cleaned.csv")
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df
```

```python
df.dtypes
```

# 5 Line chart 

 # 1. Monthly Sales Trend

```python
sales = df.groupby("order_date")["net_amount"].sum()
```

```python
plt.figure(figsize=(10,8))
plt.plot(sales.index, sales.values,color='blue')
plt.title("Monthly Sales Trend")
plt.xlabel("Order Date")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

# 2.Regional Sales

```python
plt.figure(figsize=(8,5))
region = df.groupby("region")["net_amount"].sum()
plt.plot(region.index, region.values,color='black')
plt.title("Regional Sales")
plt.xlabel("Region")
plt.ylabel("Net Amount")
plt.grid()
plt.show()
```

 # 3. Customer Age vs Spending

```python
plt.figure(figsize=(8,5))
age = df.groupby("customer_age")["net_amount"].mean()
plt.plot(age.index, age.values,color='green')
plt.title("Customer Age vs Average Spending")
plt.xlabel("Customer Age")
plt.ylabel("Average Spending")
plt.grid()
plt.show()
```

#  4. Customer Ratings

```python
plt.figure(figsize=(8,5))
rating = df.groupby("product_category")["customer_rating"].mean()
plt.plot(rating.index, rating.values,)
plt.title("Customer Ratings")
plt.xlabel("Category")
plt.ylabel("Average Rating")
plt.show()
```

# 5 Daily Sales Trend

```python
plt.figure(figsize=(8,5))
sales = df.groupby("order_date")["net_amount"].sum()
plt.plot(sales.index, sales.values,color='green')
plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.show()
```

# Bar Graph

# 1. Revenue by Payment Method

```python
data = df.groupby("payment_method")["net_amount"].sum()

plt.figure(figsize=(10, 5))
plt.bar(data.index, data.values, color="skyblue", width=0.6, edgecolor="black", label="Revenue")
plt.title("Revenue by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Revenue")
plt.legend()
plt.show()
```

# 2.Revenue by Sales Channel

```python
data = df.groupby("sales_channel")["net_amount"].sum()
plt.figure(figsize=(10, 5))
plt.bar(data.index, data.values, color="purple",width=0.6, edgecolor="black", label="Revenue")
plt.title("Revenue by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Total Revenue")
plt.legend()
plt.show()
```

# 3. Orders by Region

```python
data = df.groupby("region")["order_id"].nunique()
plt.figure(figsize=(10, 5))
plt.bar(data.index, data.values, color="lightgreen",width=0.6, edgecolor="black", label="Orders")
plt.title("Orders by Region")
plt.xlabel("Region")
plt.ylabel("Total Orders")
plt.legend()
plt.show()
```

# 4. Revenue by Marketing Campaign

```python
data = df.groupby("marketing_campaign")["net_amount"].sum()
plt.figure(figsize=(10, 5))
plt.bar(data.index, data.values, color="indigo",width=0.6, edgecolor="black", label="Revenue")
plt.title("Revenue by Marketing Campaign")
plt.xlabel("Marketing Campaign")
plt.ylabel("Total Revenue")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```
