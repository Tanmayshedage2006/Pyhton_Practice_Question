```python
def greet():
    print("Hello!") # creating a function

greet()  # call the function
```

```python
def add(a, b):  # a and b are parameters Parmaters is the value we define during the declration of function
    print(a + b)
```

```python
add(10, 20)  # 10 and 20 are arguments # Argument is actual value pass to a function
```

```python
def total(*args): # accept any number of postional argument and stored in tuple
    print(args)

total(10, 20, 30) 
```

```python
def details(**kwargs): # kwargs accept the value in key and value pair form and stored 
    print(kwargs)

details(name="Tanmay", age=20)
```
