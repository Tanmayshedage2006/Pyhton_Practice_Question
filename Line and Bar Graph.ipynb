```python
import pandas as pd

df = pd.read_csv("industry_retail_orders_dirty_1000.csv")
```

```python
import matplotlib.pyplot as plt
```

```python
df['product_category']=(df['product_category'].str.strip()).str.title()
df['product_category'].unique()
```

```python
df=df.dropna(subset=['product_category']) # subset used for specific column
```

```python
df['product_category'].unique()
```

# Line chart 

```python
sales = df.groupby('product_category')['gross_amount'].sum()
plt.figure(figsize=(8,5))
plt.plot(sales.index, sales.values, color='blue',)
plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Gross Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
orders = df['product_category'].value_counts()
plt.figure(figsize=(8,5))
plt.plot(orders.index, orders.values, color='green',linewidth=2, label='Total Orders')
plt.title('Orders by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
avg_sales = df.groupby('product_category')['gross_amount'].mean()
plt.figure(figsize=(8,5))
plt.plot(avg_sales.index, avg_sales.values, color='yellow',linewidth=2, label='Average Sales')
plt.title('Average Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Average Gross Amount')
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
quantity = df.groupby('product_category')['quantity'].sum()
plt.figure(figsize=(8,5))
plt.plot(quantity.index, quantity.values,color='green', linewidth=2, label='Total Quantity')
plt.title('Total Quantity Sold by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Quantity')
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
max_sales = df.groupby('product_category')['gross_amount'].max()
plt.figure(figsize=(8,5))
plt.plot(max_sales.index, max_sales.values, color='purple',linewidth=2, label='Maximum Sales')
plt.title('Maximum Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Maximum Gross Amount')
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

# Bar Graph
