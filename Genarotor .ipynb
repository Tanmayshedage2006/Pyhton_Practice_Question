```python
def my_generator():
    yield 10
    yield 20
    yield 30

for x in my_generator():
    print(x)
```

```python
def table():
    for i in range(1, 11):
        yield i * 2

for x in table():
    print(x)
```

```python
def numbers():
    yield 100
    yield 200
    yield 300

g = numbers()

print(next(g))
print(next(g))
print(next(g))
```
