```python
def even_odd(n):
    try:
        if n % 2 == 0:
            return "Even Number", n
        else:
            return "Odd Number", n
    except Exception as e:
        print("Error in even_odd function")
        print(e)

print(even_odd(10))
```

```python
def divide(a, b):
    try:
        return a / b
    except Exception as e:
        print("Error in divide function")
        print(e)

print(divide(10, 0))
```

```python
def square(n):
    try:
        return n ** 2
    except Exception as e:
        print("Error in square function")
        print(e)

print(square("abc"))
```

```python
def largest(a, b):
    try:
        if a > b:
            return a
        else:
            return b
    except Exception as e:
        print("Error in largest function")
        print(e)

print(largest(10, 20))
```

```python
def read_file(filename):
    try:
        with open(filename, "r") as f:
            return f.read()
    except Exception as e:
        print("Error in read_file function")
        print(e)

print(read_file("data.txt"))
```

```python
def get_element(lst, index):
    try:
        return lst[index]
    except Exception as e:
        print("Error in get_element function")
        print(e)

numbers = [10, 20, 30]
print(get_element(numbers, 5))
```
