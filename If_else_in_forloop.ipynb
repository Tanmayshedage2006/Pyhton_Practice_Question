```python
for i in range(1, 11):
    if i % 2 == 0:
        print(i, "is Even")
    else:
        print(i, "is Odd") # Checking if number is even or odd
```

```python
numbers = [10, -5, 0, 8, -2]

for num in numbers:
    if num > 0:
        print(num, "is Positive")
    elif num < 0:
        print(num, "is Negative")
    else:
        print(num, "is Zero") # checking number is postive negative or zero
```

```python
marks = [85, 42, 67, 30, 91]

for mark in marks:
    if mark >= 35:
        print(mark, "- Pass")
    else:
        print(mark, "- Fail")
```

```python
for i in range(1, 16):
    if i % 3 == 0:
        print(i, "is a multiple of 3")
    else:
        print(i, "is not a multiple of 3")
```
