```python
class person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def __repr__(self):
        return f"Person ('{self.name}',{self.age})"

p1=person("Tanmay",34)
print(p1)
```

```python
class person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def __str__(self):
        return f"{self.name} ({self.age})"
        
    def __repr__(self):
        return f"Person ('{self.name}',{self.age})"

p1=person("Tanmay",23)
print(p1)
print(repr(p1))
```

 # __eq__

```python
class person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

p1=person("Tanmay",15)
p2=person("Tanmay",15)

print(p1 == p2)  # memory mahe velgele velegle yete
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def __eq__(self,other):
        return self.name == other.name and self.age == other.age

p1=Person("Tanmay",15)
p2=Person("Tanmay",15)


print(p1 == p2)
```

# add

```python
class person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __add__(self,other):
        return self.age + other.age

p1=person("Tanmay",15)
p2=person("Tanmay",15)

print(p1 + p2)
```

 # __len__

```python
class company:
    def __init__(self, employees):
        self.employees = employees

    def __len__(self):
        return len(self.employees)

c1 = company(["Tanmay", "Gaurav", "Sarthak"])

print(len(c1))
```

# __IT__

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __lt__(self, other):
        return self.age < other.age

p1 = Person("Tanmay", 15)
p2 = Person("Tanmay", 15)

print(p1 < p2)
```
