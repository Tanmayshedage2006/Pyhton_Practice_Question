```python
num = 1

while num <= 10:
    print(num) # Checking if it's prime number
    num += 1
```

```python
num = 10

while num >= 1:
    print(num)
    num -= 1   # Printing to the reverse the series
```

```python
num = 1

while num <= 10:
    if num % 2 == 0:
        print(num)
    num += 1  # Checking even or odd number
```

```python
num = 1
total = 0

while num <= 10:
    total += num
    num += 1

print("Sum =", total)
```
