```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("industry_retail_orders_dirty_1000.csv")
plt.figure(figsize=(8,5))
plt.hist(df['customer_age'], bins=10, color='skyblue', edgecolor='black')
plt.title("Distribution of Customer Age")
plt.xlabel("Customer Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['quantity'], bins=10, color='yellow', edgecolor='black')
plt.title("Distribution of Product Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['net_amount'], bins=15, color='lightblue',edgecolor='black')
plt.title("Distribution of Net Amount")
plt.xlabel("Net Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.hist(df['customer_age'], bins=5, color='skyblue',
         edgecolor='black')
plt.title("Customer Age - 5 Bins")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.subplot(1,3,2)
plt.hist(df['customer_age'], bins=10, color='orange',
         edgecolor='black')
plt.title("Customer Age - 10 Bins")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.subplot(1,3,3)
plt.hist(df['customer_age'], bins=20, color='lightgreen',
         edgecolor='black')
plt.title("Customer Age - 20 Bins")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```
