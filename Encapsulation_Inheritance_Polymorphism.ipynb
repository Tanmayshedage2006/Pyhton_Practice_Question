# Encapsultion

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks    # Private variable

    def show_marks(self):
        print("Marks:", self.__marks)

    def update_marks(self, marks):
        self.__marks = marks


s = Student("Tanmay", 85)

print("Name:", s.name)
s.show_marks()                        # there are three parts into it :- private ,protect,public


s.update_marks(95)
s.show_marks()
```

# Inheritance

```python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


d = Dog()

d.eat()
d.bark()  # parent class --> child class
```

# Polymorphism

```python
class Animal:
    def sound(self):
        print("Animal makes sound")


class Dog(Animal):
    def sound(self):
        print("Dog barks")


class Cat(Animal):
    def sound(self):
        print("Cat meows")


d = Dog()
c = Cat()

d.sound()
c.sound()
```
